Exercise 1: Prime Numbers Write a Python program that checks whether a given number is prime or not. A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself.

```python
def is_prime(n):
    if n < 2: return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

print(f"is_prime(12) = {is_prime(12)}")  # True
print(f"is_prime(19) = {is_prime(19)}")  # False
```

Exercise 2: Product of Random Numbers Develop a Python program that generates two random numbers and asks the user to enter the product of these numbers. The program should then check if the user's answer is correct and display an appropriate message.

```python
import random

def product_doll():

    num1 = random.randint(1, 10)
    num2 = random.randint(10, 20)
    correct_answer = num1 * num2

    # Ask the user for the product
    print(f"Enter the product doll  {num1} * {num2}?")

    try:
        user_answer = int(input("product doll: "))

        # Check if the answer is correct
        if user_answer == correct_answer:
            print(f"Correct. right answer {correct_answer} .")
        else:
            print(f" Wrong. The correct answer was {correct_answer}.")
    except ValueError:
        print("Invalid input. Please enter a whole number.")

# Run the program
product_doll()
```

Squares of Even/Odd Numbers Create a Python script that prints the squares of all even or odd numbers within the range of 100 to 200. Choose either even or odd numbers and document your choice in the code.

```python
# square of each even/odd number between 100 to 200.

print("Squares of even/odd numbers from 100 to 200:")

for num in range(100, 201):
    if num % 2 == 0:
        square = num ** 2
        print(f"{num} squared is {square}")
```

✅ Problem Statement (Question): Hospital Billing System A hospital charges its patients based on the following:

1.Room charges per day: ₹2,000

2.Doctor consultation fee (flat): ₹1,500

3.Lab test charges: Based on the number of tests taken. Each test costs ₹300

4.Medicine charges: Total cost of medicines provided

5.Discount: If the total bill (before discount) exceeds ₹10,000, a 10% discount is applied

Write a Python program to calculate the final bill for a patient who:

1.Stayed for 4 days

2.Had 3 lab tests

3.Had medicine charges of ₹2,400

```python
def hospital_bill(days,lab_tests,medicine_charges):
    total_amount = 0
    total_amount = days *2000 + 1500 + 300*lab_tests + medicine_charges
    if total_amount >10000:
        total_amount = total_amount-( total_amount * 0.10)
    return total_amount
print("Final Bill : ",hospital_bill(4,3,2400))
```
